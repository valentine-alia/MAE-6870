```python
import numpy as np
import sympy as sym
from itertools import combinations

def multiplicative_compound(A, k):
    """
    Return the kth multiplicative compound matrix C_k(A).

    C_k(A) is the matrix of all k x k minors of A, with
    rows and columns indexed lexicographically by k-element
    subsets of the row and column indices.

    Parameters
    ----------
    A : sympy.Matrix
        An m x n matrix.
    k : int
        Size of the minors.

    Returns
    -------
    sympy.Matrix
        A matrix of size binomial(m, k) x binomial(n, k).
    """
    A = sym.Matrix(A)
    m, n = A.shape

    if not 0 <= k <= min(m, n):
        raise ValueError("k must satisfy 0 <= k <= min(A.shape)")

    row_subsets = list(combinations(range(m), k))
    col_subsets = list(combinations(range(n), k))

    return sym.Matrix([
        [
            A.extract(rows, cols).det()
            for cols in col_subsets
        ]
        for rows in row_subsets
    ])
```

```python
def additive_compound(A, k):
    n = A.rows
    eps = sym.symbols('eps')

    # I + eps*A
    B = sym.eye(n) + eps * A

    # Multiplicative k-compound
    Bk = multiplicative_compound(B, k)

    # Differentiate entrywise and evaluate at eps = 0
    return Bk.applyfunc(lambda x: sym.diff(x, eps).subs(eps, 0))
```
