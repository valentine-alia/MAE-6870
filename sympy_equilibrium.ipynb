\begin{equation}
0 = A(\frac{C}{1+ (C/k)^n}) + DC - \gamma
\end{equation}


Where $A = \frac{\omega_C}{\alpha_C}(\alpha_C+\alpha_B+\alpha_T + \gamma + \gamma \frac{\alpha_T}{\beta}), D = \gamma(1 + \frac{\omega_C\alpha_B}{\omega_B\alpha_C})$

```python
n = 4
A = omegaB/alphaC * (alphaC + alphaB + alphaT + gamma + gamma*alphaT/beta)
D = gamma * (1 + (omegaC * alphaB)/(omegaB*alphaC))
satch = 1 + (C/k)**n
p = A*C + satch*(D*C - gamma)
sym.roots(p,C)
```
